<div class="guide" dir="rtl" align="right" style="direction:rtl;text-align:right;">
  <h2 style="direction:rtl;text-align:right;">ابزار خواندن API</h2>
  <p>سلول پایین را یک بار اجرا کن. تابع <code>get_json</code> همان کاری را می‌کند که مرورگر در پنل Network می‌کند: آدرس را می‌زند و JSON را به دیکشنری یا لیست پایتون تبدیل می‌کند.</p>
  <p>اگر خواستی خودت آدرس تازه‌ای پیدا کنی: سایت را باز کن، ابزار توسعه‌دهنده، زبانه Network، فیلتر Fetch/XHR، صفحه را رفرش کن، درخواست JSON قیمت را باز کن و آدرس را کپی کن.</p>
</div>


```python
try:
    import requests
except ModuleNotFoundError:
    raise SystemExit("کتابخانه requests نصب نیست. در ترمینال بنویس: pip install requests")

def get_json(url):
    response = requests.get(
        url,
        headers={
            "User-Agent": "Mozilla/5.0",
            "Accept": "application/json",
        },
        timeout=30,
    )
    response.raise_for_status()
    return response.json()

print("ابزار API آماده است. سؤال‌ها از همین get_json استفاده می‌کنند.")
```
