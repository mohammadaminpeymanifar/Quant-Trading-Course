  <p>درصدها را با دو رقم اعشار چاپ کن: <code>print(f"{value:.2f}")</code>. نام نمادها را همان‌طور که در داده آمده چاپ کن، با حروف کوچک و بزرگ خودشان.</p>

```python
# این سلول را اول از همه اجرا کن تا فونت وزیر و استایل سؤال‌ها فعال شود.
from IPython.display import HTML, display
from pathlib import Path

def find_style():
    seeds = [Path.cwd(), *list(Path.cwd().parents)[:4]]
    for seed in seeds:
        for rel in (
            Path("fonts") / "notebook_style.html",
            Path("Student") / "fonts" / "notebook_style.html",
        ):
            candidate = seed / rel
            if candidate.exists():
                return candidate
    matches = sorted(Path.cwd().glob("**/fonts/notebook_style.html"))
    if matches:
        return matches[0]
    return None

style_path = find_style()
if style_path is None:
    print("هشدار: fonts/notebook_style.html پیدا نشد.")
    print("نوت‌بوک را از پوشه همان آزمون باز کن تا پوشه fonts کنارش باشد.")
else:
    print("استایل از این مسیر خوانده شد:", style_path)
    display(HTML(style_path.read_text(encoding="utf-8")))
    print("فونت وزیر و رنگ سؤال‌ها اعمال شد.")
```

  <p><b>بازده کل پنجره:</b> (آخرین قیمت − اولین قیمت) / اولین قیمت × ۱۰۰. این بازدهِ یک روز آخر نیست؛ کل مسیر را در یک عدد خلاصه می‌کند.</p>
  <p><b>مومنتوم در این دفتر:</b> همان بازده کل پنجره. سهمی که این عددش بزرگ‌تر باشد، در این تعریف مومنتوم قوی‌تری دارد. این فقط یک قاعده آموزشی رتبه‌بندی است.</p>
  <p><b>روز سبز:</b> روزی که قیمت پایانی از مشاهده قبلی بیشتر باشد. روز اول روز سبز نیست، چون دیروزی داخل لیست ندارد.</p>
  <p><b>میانگین متحرک ساده سه‌روزه:</b> میانگین سه قیمت متوالی. هر بار که پنجره یک روز جلو می‌رود، قدیمی‌ترین قیمت از میانگین بیرون می‌افتد و قیمت تازه داخل می‌آید. اگر آخرین قیمت از میانگین سه قیمت آخر بزرگ‌تر باشد قاعده این دفتر BUY است وگرنه SELL. برابر بودن هم SELL است.</p>
  <p><b>بازده بخش:</b> میانگین بازده کل سهم‌هایی که در آن بخش هستند. بخش، خودش یک دارایی جدا نیست.</p>

  <p>ابزار پایتون این آزمون: <code>list</code> · <code>dict</code> · <code>tuple</code> · <code>set</code> · متدهای <code>string</code> · <code>for</code> · <code>if</code></p>
  <ul>
