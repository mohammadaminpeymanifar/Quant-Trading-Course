```python
# این سلول را اول از همه اجرا کن تا فونت وزیر و استایل سؤال‌ها فعال شود.
from IPython.display import HTML, display
from pathlib import Path

def find_style():
    seeds = [Path.cwd(), *list(Path.cwd().parents)[:4]]
    for seed in seeds:
        for rel in (
            Path("fonts") / "notebook_style.html",
            Path("Student") / "fonts" / "notebook_style.html",
        ):
            candidate = seed / rel
            if candidate.exists():
                return candidate
    matches = sorted(Path.cwd().glob("**/fonts/notebook_style.html"))
    if matches:
        return matches[0]
    return None

style_path = find_style()
if style_path is None:
    print("هشدار: fonts/notebook_style.html پیدا نشد.")
    print("نوت‌بوک را از پوشه همان آزمون باز کن تا پوشه fonts کنارش باشد.")
else:
    print("استایل از این مسیر خوانده شد:", style_path)
    display(HTML(style_path.read_text(encoding="utf-8")))
    print("فونت وزیر و رنگ سؤال‌ها اعمال شد.")
```

  <p><b>NAV:</b> ارزش هر واحد صندوق در پایان روز. در این دفتر به‌جای قیمت سهم، همین سری را می‌خوانی.</p>
  <p><b>سقف جاری:</b> بیشترین NAV از روز صفر تا همان روز، شامل خود آن روز. سقف هیچ‌وقت با افت قیمت پایین نمی‌آید.</p>
  <p><b>افت از سقف همان روز:</b> (سقف جاری − NAV امروز) / سقف جاری × ۱۰۰. روی سقف، این عدد صفر است.</p>
  <p><b>بیشترین افت:</b> بزرگ‌ترین عدد همین سری افت‌ها. صندوقی که این عددش بزرگ‌تر باشد، در این پنجره افت دردناک‌تری از سقف دیده، حتی اگر آخر کار برگردد.</p>
  <p><b>روز بازیابی:</b> روزی که NAV از دیروز بیشتر شده، ولی هنوز به سقفِ قبل از امروز نرسیده. روزی که خودش سقف تازه می‌سازد روز بازیابی نیست؛ آن روز، روند به اوج جدید رسیده است.</p>
