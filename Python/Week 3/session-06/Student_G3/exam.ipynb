```python
# این سلول را اول از همه اجرا کن تا فونت وزیر و استایل سؤال‌ها فعال شود.
from IPython.display import HTML, display
from pathlib import Path

def find_style():
    seeds = [Path.cwd(), *list(Path.cwd().parents)[:4]]
    for seed in seeds:
        for rel in (
            Path("fonts") / "notebook_style.html",
            Path("Student") / "fonts" / "notebook_style.html",
        ):
            candidate = seed / rel
            if candidate.exists():
                return candidate
    matches = sorted(Path.cwd().glob("**/fonts/notebook_style.html"))
    if matches:
        return matches[0]
    return None

style_path = find_style()
if style_path is None:
    print("هشدار: fonts/notebook_style.html پیدا نشد.")
    print("نوت‌بوک را از پوشه همان آزمون باز کن تا پوشه fonts کنارش باشد.")
else:
    print("استایل از این مسیر خوانده شد:", style_path)
    display(HTML(style_path.read_text(encoding="utf-8")))
    print("فونت وزیر و رنگ سؤال‌ها اعمال شد.")
```

  <h2 style="direction:rtl;text-align:right;">مفاهیم این دفتر</h2>
  
  <p><b>ارزش ذاتی آموزشی این دفتر:</b> اونس جهانی × نرخ دلار / ۱۰۰۰. تقسیم بر ۱۰۰۰ فقط مقیاس را کوچک کرده تا عددها در تمرین جا شوند. در بازار واقعی وزن سکه و تبدیل عیار هم وارد فرمول می‌شود؛ اینجا آن را عمداً ساده کرده‌ایم تا خود مفهوم حباب دیده شود.</p>
  <p><b>حباب درصدی:</b> (قیمت بازار − ارزش ذاتی) / ارزش ذاتی × ۱۰۰. مثبت یعنی بازار بالای ارزش ذاتی معامله می‌شود.</p>
  <p><b>نرخ بهره حقیقی تقریبی:</b> نرخ اسمی سپرده منهای تورم همان سال. این تقریب درسی است، نه فرمول دقیق فیشِر. اگر نتیجه منفی باشد، سود اسمی سپرده از تورم همان سال عقب مانده است.</p>
  <p><b>شکاف دلار:</b> (نرخ آزاد − نرخ نیما) / نرخ نیما × ۱۰۰. مثبت یعنی آزاد گران‌تر از نیما است.</p>

  <p>ابزار پایتون این آزمون: <code>list</code> · <code>dict</code> · <code>tuple</code> · <code>set</code> · متدهای <code>string</code> · <code>for</code> · <code>if</code></p>
  <ul>
    <li>لیست: <code>items[0]</code> اول، <code>items[-1]</code> آخر، <code>items[-3:]</code> سه عضو آخر.</li>
    <li>دیکشنری: <code>d[key]</code> خواندن، <code>d[key] = v</code> نوشتن، <code>for k, v in d.items()</code> پیمایش.</li>
    <li>تاپل: <code>a, b = pair</code> باز کردن. در حلقه هم می‌شود تاپل را باز کرد.</li>
    <li>مجموعه: عضو تکراری ندارد. اشتراک <code>&amp;</code> · تفاضل <code>-</code>. برای چاپ، <code>sorted</code> بگیر تا ترتیب ثابت باشد.</li>
    <li>رشته: <code>strip</code> · <code>upper</code> · <code>title</code> · <code>split</code> · <code>join</code> · <code>replace</code> · <code>isdigit</code> · <code>startswith</code>.</li>
    <li>اگر کلیدی هنوز در دیکشنری نیست، اول با <code>if key not in d</code> مقدار اولیه بگذار، بعد جمع کن.</li>
